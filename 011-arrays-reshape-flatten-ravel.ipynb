### Reshaping arrays

```python
import numpy as np
```

```python
arr1=np.arange(1,17)
arr1
```

```python
arr2=np.reshape(arr1,(4,4))
arr2
```

```python
arr3=np.reshape(arr1,(8,2))
arr3
```

```python
# arr4=np.reshape(arr1,(3,4))
```

```python
arr5=np.reshape(arr1,(4,2,2))
print(arr5)
print("Dimensions of fifth_arr is ",arr5.ndim)
```

```python
arr6=np.array([[1,2],[3,4],[5,6]])
arr6
```

```python
arr7_to_flatten=np.reshape(arr6,-1)
arr7_to_flatten
```

```python
arr8_to_flatten=arr6.flatten()
print("arr8_to_flatten: ",arr8_to_flatten)
arr9_to_ravel=arr6.ravel()
print("arr9_to_ravel: ",arr9_to_ravel)
```

```python
arr8_to_flatten[0]=10
```

```python
arr9_to_ravel[0]=20
```

```python
print("eighth_arr_flat: ",arr8_to_flatten)
print("arr9_to_ravel: ",arr9_to_ravel)
print("arr6: ",arr6)
```
