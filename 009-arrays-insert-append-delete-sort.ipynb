### Adding, removing and sorting elements

```python
import numpy as np
```

```python
arr1=np.array([1, 2, 4, 5, 6])
arr1
```

```python
arr1_new=np.insert(arr1,1,3)
arr1_new
```

```python
arr2=np.array([1,2,3])
arr2
```

```python
arr2_new=np.append(arr2,4)
arr2_new
```

```python
arr3=np.array([1,2])
arr3
```

```python
arr3_to_delete=np.delete(arr3,1)
arr3_to_delete
```

```python
integers_1d_arr=np.random.randint(0,10,10)
integers_1d_arr
```

```python
print(np.sort(integers_1d_arr))
```

```python
integers_2d_arr=np.array([[-1, -2, 0,7, 2], [4, 9, -2, 8, 9]])
integers_2d_arr
```

```python
print(np.sort(integers_2d_arr))
```

```python
colors_arr=np.array(['pink','orange','gray','white','black','yellow','sky blue'])
colors_arr
```

```python
print(np.sort(colors_arr))
```
