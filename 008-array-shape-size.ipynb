###  Finding the shape and size of an array

```python
import numpy as np
```

```python
arr1=np.arange(60)
arr1
```

```python
arr2=np.linspace((1,2),(10,60),10)
arr2
```

```python
arr3=np.full((2,2,2),10)
arr3
```

```python
np.shape(arr1)
```

```python
np.shape(arr2)
```

```python
np.shape(arr3)
```

```python
np.size(arr1)
```

```python
np.size(arr2)
```

```python
np.size(arr3)
```
