### Copies and views

```python
import numpy as np
```

```python
students_ids_num=np.array([1111,1112,1113,1114,1115,1116,1117,1118])
students_ids_num
```

```python
students_ids_num_reg=students_ids_num
print("ID's of students_ids_num",id(students_ids_num))
print("ID's of students_ids_num_reg",id(students_ids_num_reg))
```

```python
students_ids_num_reg[1]=1122
print(students_ids_num)
print(students_ids_num_reg)
```

```python
students_ids_num_cp=students_ids_num.copy()
```

```python
print(students_ids_num_cp)
```

```python
print(students_ids_num_cp==students_ids_num)
```

```python
print ("ID's of students_ids_num",id(students_ids_num))
print("ID's of students_ids_num_cp",id(students_ids_num_cp))
```

```python
students_ids_num[0]=1000
print ("Original ID's: ", students_ids_num)
print("Copy of ID's: ",students_ids_num_cp)
```

```python
students_ids_num_v=students_ids_num.view()
```

```python
students_ids_num_v[0]=2000
print("Original ID's: ", students_ids_num)
print("Backed (Attached) view of ID's: ",students_ids_num_v)
```

```python
print(students_ids_num_cp.base)
print(students_ids_num_v.base)
```
