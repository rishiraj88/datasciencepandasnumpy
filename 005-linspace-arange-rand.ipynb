### Intrisic NumPy array creation

```python
import numpy as np
```

```python
integers_arr1=np.arange(10)
integers_arr1
```

```python
integers_arr2=np.arange(100,200)
integers_arr2
```

```python
integers_arr3=np.arange(100,201,2)
integers_arr3
```

```python
floats_arr1=np.linspace(10,50)
floats_arr1
```

```python
floats_arr2=np.linspace(10,50,5)
floats_arr2
```

```python
rand_arr1=np.random.rand(5)
rand_arr1
```

```python
rand_arr2=np.random.rand(3,3)
rand_arr2
```

```python
rand_arr3=np.random.randint(0,49,30)
rand_arr3
```
