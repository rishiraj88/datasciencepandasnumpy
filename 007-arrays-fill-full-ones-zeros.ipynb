### Creating arrays with constants

```python
import numpy as np
```

```python
zero_arr1=np.zeros(5)
zero_arr1
```

```python
zero_arr2=np.zeros((4,5))
zero_arr2
```

```python
ones_arr1=np.ones(6)
ones_arr1
```

```python
ones_arr2=np.ones((7,8))
ones_arr2
```

```python
ones_arr3=np.ones((4,5),dtype=int)
ones_arr3
```

```python
fill_arr1=np.empty(10,dtype=int)
fill_arr1.fill(12)
fill_arr1
```

```python
full_arr1=np.full(5,10)
full_arr1
```

```python
full_arr2=np.full((4,5),8)
full_arr2
```
